**Chapter 3 : Strings** #Strings are unmutable, can't change


```python
# Intro to Strings ; Positive String count start with 0,2,3..eg. Harrry (ASC)and Neg Start with -1,-2,-3...(DESC) eg. yrraH

name = "Harry"
nameshort = name[0:3] # start from index 0 all the way till 3 (excluding 3)
print(nameshort)
character1 = name[1]
print(character1)
```

```python
# Negative Slicing
name = "Harry"

print(name[0:3])

print(name[-4: -1])
print(name[1:4])

print(name[:4]) # is same as print(name[0:4])
print(name[1:]) # is same as print(name[1:5])
print(name[1:5])
```

```python
#STR Functions

name = "harry"

print(len(name))
print(name.endswith("rry"))
print(name.startswith("ha"))
print(name.capitalize())
print(name.lower())
```

```python
#escape_seq
a = 'Harry is a good boy\nbut not a bad \'boy\'' # /n use for New Line

print(a)
```

**Chapter 4 : LIST**


```python
# LIST

friends = ["Apple", "Orange", 5, 345.06, False, "Aakash", "Rohan"]

print(friends[0])
friends[0] = "Grapes" # Unlike Strings lists are mutable, can change

print(friends[0])
print(friends[1:4])
```

```python
# List Method

friends = ["Apple", "Orange", 5, 345.06, False, "Aakash", "Rohan"]
print(friends)
friends.append("Harry")
print(friends)

l1 = [1, 34,62, 2, 6, 11]
#l1.sort() # Sort ASC
#l1.reverse() # DESC
# l1.insert(2, 333333) # Insert 333333 such that its index in the list is 3
value = l1.pop(3)
print(value)
print(l1)
```

```python
# TUPLE ; Can't Change Tuple

a = (1,45,342,3424,False, "Rohan", "Shivam")
print(a)
print(type(a))
```

```python
# TUPLE METHOD

a = (1,45,342,3424,False, 45, "Rohan", "Shivam")
print(a)

no = a.count(45)
print(no)

i = a.index(3424)
print(i)

print(len(a))
```

**Chapter 5 : Dictionary & Sets**





```python
# DICT : Unordered, Mutable, Indexed, Can't Contain duplicate keys

d = {} # Empty dictionary
marks = {
    "Harry": 100,
    "Shubham": 56,
    "Rohan": 23
}
print(marks["Harry"])  # print(marks, type(marks))
print(marks["Shubham"])
```

```python
# DICT METHOD

marks = {
    "Harry": 100,
    "Shubham": 56,
    "Rohan": 23,
    0: "Harry"
}

print(marks.items())
print(marks.keys())
print(marks.values())
marks.update({"Harry": 99, "Renuka": 100}) # If value found in dict than update if not found than add eg. Renuka : 100
print(marks)

print(marks.get("Harry2")) # Prints None
#print(marks["Harry2"]) # Returns an error
```

```python
# SET : Unordered ,Unindexed, can't change sets & contain duplicate values

e = set() # Dont use s = {} as it will create an empty dictionary
s = {1, 5, 32, 54,5, 5, 5} # Set can print repeated number is one time only

print(s)
```

```python
# SET METHODS

s = {1, 5, 32, 54,5, 5, 5, "Harry"}

print(s, type(s))

s.add(566)
print(s, type(s))
s.remove(1)
print(s, type(s))
```

```python
# Set_Union_Intersection

s1 = {1, 45, 6, 78}
s2 = {7, 8, 1, 78}

print(s1.union(s2))
print(s1.intersection(s2))
```
